```python
import os
import sys

# Registrar rutas de librerías de Conda y del sistema
conda_env = r"C:\Users\joral\anaconda3\envs\sde_tf"
dll_dirs = [
    os.path.join(conda_env, "Library", "bin"),
    os.path.join(conda_env, "bin"),
    r"C:\Windows\System32"
]

for p in dll_dirs:
    if os.path.exists(p):
        os.add_dll_directory(p)
        os.environ["PATH"] = p + os.pathsep + os.environ.get("PATH", "")

import tensorflow as tf

gpus = tf.config.list_physical_devices('GPU')
print("GPUs detectadas en el Notebook:", len(gpus))
if gpus:
    print("Dispositivo:", gpus[0].name)
```
