```python
import sys
import torch

print("=" * 60)
print(f"Python: {sys.version.split()[0]}")
print(f"PyTorch versión: {torch.__version__}")
print("=" * 60)

# 1. Comprobación de disponibilidad de CUDA
cuda_disponible = torch.cuda.is_available()
print(f"¿CUDA disponible?: {cuda_disponible}")

if not cuda_disponible:
    print("\n[ERROR] PyTorch no detecta la GPU.")
    print("Posibles causas:")
    print(" - Tienes instalada la versión de PyTorch 'cpu-only'.")
    print(" - Falta actualizar el driver de NVIDIA en Windows.")
    sys.exit(1)

# 2. Información del hardware
n_gpus = torch.cuda.device_count()
print(f"Número de GPUs detectadas: {n_gpus}")

for i in range(n_gpus):
    nombre = torch.cuda.get_device_name(i)
    vram_gb = torch.cuda.get_device_properties(i).total_memory / (1024**3)
    compute_cap = torch.cuda.get_device_capability(i)
    print(f" -> GPU [{i}]: {nombre} | VRAM: {vram_gb:.2f} GB | Capacidad: {compute_cap}")

# 3. Test de cómputo matricial directo en VRAM
try:
    device = torch.device("cuda:0")
    a = torch.randn(5000, 5000, device=device)
    b = torch.matmul(a, a.t())
    torch.cuda.synchronize()  # Sincronización para asegurar la ejecución
    print("\n[OK] Cómputo matricial en GPU realizado con éxito.")
    print("=> Entorno sde_torch_env listo y validado en GPU.")
except Exception as e:
    print(f"\n[ERROR] Falló el cálculo matricial en GPU: {e}")
    sys.exit(1)
```
